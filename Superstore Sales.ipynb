Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

# Plot settings
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Load the Superstore dataset
df = pd.read_csv("superstore final.csv", encoding="latin1")

print("Dataset loaded successfully!")
```

Data Types

Missing Values

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
df = df.drop_duplicates()

print("Duplicate rows after removal:", df.duplicated().sum())
```

Descriptive Statistics

```python
df.describe()
```

Mean

```python
numeric_columns = df.select_dtypes(include=np.number).columns

mean_values = df[numeric_columns].mean()

print("Mean:")
print(mean_values)
```

Medain

```python
median_values = df[numeric_columns].median()

print("Median:")
print(median_values)
```

Mode

```python
mode_values = df[numeric_columns].mode().iloc[0]

print("Mode:")
print(mode_values)
```

SD

```python
std_values = df[numeric_columns].std()

print("Standard Deviation:")
print(std_values)
```

Convert Order Date

```python
# Convert Order Date to datetime format
if 'Order Date' in df.columns:
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')

print(df['Order Date'].dtype)
```

Check Date Range

```python
print("Minimum Order Date:", df['Order Date'].min())
print("Maximum Order Date:", df['Order Date'].max())
```

Create Year and Month Columns

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Month Name'] = df['Order Date'].dt.month_name()
```

Create Quarter

```python
df['Quarter'] = df['Order Date'].dt.to_period('Q').astype(str)

df[['Order Date', 'Year', 'Month', 'Month Name', 'Quarter']].head()
```

Check Sales Column

```python
print(df[['Sales']].head())
```

Total Sales

```python
total_sales = df['Sales'].sum()

print(f"Total Sales: ${total_sales:,.2f}")
```

Total Profit

```python
if 'Profit' in df.columns:
    total_profit = df['Profit'].sum()
    print(f"Total Profit: ${total_profit:,.2f}")
```

Total Quantity

```python
if 'Quantity' in df.columns:
    total_quantity = df['Quantity'].sum()
    print("Total Quantity Sold:", total_quantity)
```

Number of Orders

```python
if 'Order ID' in df.columns:
    total_orders = df['Order ID'].nunique()
    print("Total Unique Orders:", total_orders)
```

📈 MONTHLY SALES ANALYSIS

Monthly Sales

```python
monthly_sales = (
    df.groupby(df['Order Date'].dt.to_period('M'))['Sales']
    .sum()
)

monthly_sales.head()
```

Monthly Sales Line Chart

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker='o'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Find Highest Sales Month

```python
highest_month = monthly_sales.idxmax()
highest_month_sales = monthly_sales.max()

print("Highest Sales Month:", highest_month)
print(f"Sales: ${highest_month_sales:,.2f}")
```

Find Lowest Sales Month

```python
lowest_month = monthly_sales.idxmin()
lowest_month_sales = monthly_sales.min()

print("Lowest Sales Month:", lowest_month)
print(f"Sales: ${lowest_month_sales:,.2f}")
```

## Observation – Monthly Sales

The monthly sales analysis shows variations in sales performance
throughout the year. The highest-performing month should be examined
to understand the factors contributing to increased sales, while
low-performing months may require additional promotional activities.

📊 QUARTERLY SALES

```python
quarterly_sales = (
    df.groupby('Quarter')['Sales']
    .sum()
)

quarterly_sales
```

Quarterly Line Chart

```python
plt.figure(figsize=(10, 6))

plt.plot(
    quarterly_sales.index,
    quarterly_sales.values,
    marker='o'
)

plt.title('Quarterly Sales Trend')
plt.xlabel('Quarter')
plt.ylabel('Sales')

plt.tight_layout()
plt.show()
```

Best Quarter

```python
best_quarter = quarterly_sales.idxmax()
best_quarter_sales = quarterly_sales.max()

print("Highest Sales Quarter:", best_quarter)
print(f"Sales: ${best_quarter_sales:,.2f}")
```

## Observation – Quarterly Sales

The quarterly analysis helps identify the strongest and weakest
business periods. The highest-performing quarter can be used to
understand seasonal demand and improve future sales planning.

👥 CUSTOMER ANALYSIS

Customer Count

```python
if 'Customer ID' in df.columns:
    unique_customers = df['Customer ID'].nunique()
    print("Number of Unique Customers:", unique_customers)
```

Sales by Customer Segment

```python
if 'Segment' in df.columns:
    
    segment_sales = (
        df.groupby('Segment')['Sales']
        .sum()
        .sort_values(ascending=False)
    )
    
    print(segment_sales)
```

Segment Sales Chart

```python
if 'Segment' in df.columns:

    plt.figure(figsize=(8, 5))

    segment_sales.plot(kind='bar')

    plt.title('Sales by Customer Segment')
    plt.xlabel('Customer Segment')
    plt.ylabel('Sales')
    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()
```

## Observation – Customer Segment

The customer segment analysis shows which customer groups contribute
the most sales. The highest-performing segment should receive
continued attention through targeted marketing and customer
retention strategies.

🛍️ PRODUCT ANALYSIS

Top 10 Products

```python
top_products = (
    df.groupby('Product Name')['Quantity']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

Top 10 Products Chart

```python
plt.figure(figsize=(12, 7))

top_products.sort_values().plot(kind='barh')

plt.title('Top 10 Best-Selling Products')
plt.xlabel('Quantity Sold')
plt.ylabel('Product Name')

plt.tight_layout()
plt.show()
```

Best-Selling Product

```python
best_product = top_products.idxmax()
best_product_quantity = top_products.max()

print("Best-Selling Product:", best_product)
print("Quantity Sold:", best_product_quantity)
```

## Observation – Top Products

The top-selling product analysis identifies products with the highest
demand. These products should be monitored carefully to maintain
sufficient inventory and avoid stock shortages.

🏷️ CATEGORY ANALYSIS

Revenue by Category

```python
category_sales = (
    df.groupby('Category')['Sales']
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

Category Sales Chart

```python
plt.figure(figsize=(9, 6))

category_sales.plot(kind='bar')

plt.title('Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

Highest Revenue Category

```python
highest_category = category_sales.idxmax()
highest_category_sales = category_sales.max()

print("Highest Revenue Category:", highest_category)
print(f"Sales: ${highest_category_sales:,.2f}")
```

## Observation – Product Category

The category analysis identifies which product categories generate
the highest revenue. The company can prioritize high-performing
categories while developing strategies to improve weaker categories.

🔥 PROFIT ANALYSIS

Profit by Category

Sales vs Profit

🔥 CORRELATION HEATMAP

```python
numeric_df = df.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

correlation_matrix
```

Heatmap

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap')

plt.tight_layout()
plt.show()
```

## Observation – Correlation Heatmap

The correlation heatmap shows the relationships between numerical
variables. Strong positive or negative correlations can help identify
variables that are associated with sales, profit, quantity, or other
business measures.
